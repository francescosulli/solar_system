# 01 Build Dataset (DE440 strict)

Configurazione consigliata: finestra 2010-2030 con passo 3h.


```python
from pathlib import Path
import sys

_candidates = [Path.cwd().resolve(), Path.cwd().resolve().parent, Path.cwd().resolve().parent.parent]
_PROJECT_ROOT = next((p for p in _candidates if (p / 'src' / 'solsys_emulator').exists()), None)
if _PROJECT_ROOT is None:
    raise RuntimeError('Impossibile trovare la project root con src/solsys_emulator')

_SRC = _PROJECT_ROOT / 'src'
if str(_SRC) not in sys.path:
    sys.path.insert(0, str(_SRC))

print('Project root:', _PROJECT_ROOT)
print('Python executable:', sys.executable)

```

```python
from solsys_emulator.config import DEFAULT_BODIES, DEFAULT_DATASET_PATH, DEFAULT_KERNEL_PATH
from solsys_emulator.de440_dataset import build_dataset, find_local_kernel, save_dataset

# Strict mode: serve un kernel DE440/DE441 locale.
start = '2010-01-01T00:00:00'
end = '2030-01-01T00:00:00'
step_seconds = 10800.0  # 3 ore

kernel = find_local_kernel([DEFAULT_KERNEL_PATH])
if kernel is None:
    raise RuntimeError('Nessun kernel DE440/DE441 trovato in data/.')

dataset = build_dataset(
    start_time=start,
    end_time=end,
    step=step_seconds,
    bodies=DEFAULT_BODIES,
    kernel_path=kernel,
)

print('States shape:', dataset['states'].shape)
print('Num time points:', len(dataset['times_seconds']))
print('Source:', dataset['metadata']['sample_source'])
print('Frame:', dataset['metadata']['frame'], '| Timescale:', dataset['metadata']['timescale'])
print('Sampling step [hours]:', step_seconds / 3600.0)

save_path = save_dataset(DEFAULT_DATASET_PATH, dataset)
print('Dataset salvato in:', save_path)
```
